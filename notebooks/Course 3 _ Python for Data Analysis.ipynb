All of these operations are commonly named ETL (Extract, Transform and Load). ETL is a very important process in Data project and can sometimes takes up to 80% of the time of a project..

At the end of this course, you will be able to:

- **EXTRACT**: Load data from multiples sources: Flat files (Excel, CSV, etc.) and Databases (MSSQL)
- **TRANSFORM**: Realize simple and advanced manipulation of the data covering concatenating, merging, standardize, encode a dataset
- **LOAD**: Export your data in multiple format (Excel, CSV)

# 0 - Presentation of the case study


You will be working for the Head of the Emergency Department (ED) of a big university hospital. They would like to make an analysis of the **arrivals of patients in the emergency room** in order to extract meaningful insight that could help the management to plan staffing, beds and opening hours.

The hospital information system (DSI) sent you all the data:

- The ED Visit tables that were given in 3 parts (because the extract was too large)
- The Patient table containing all the patient information
- A reference table giving the French department and its associated region

Below you can find a representation of data that were sent. <br>

| Table                 | File                                                             | Key columns                                                                                                                |
| --------------------- | ---------------------------------------------------------------- | -------------------------------------------------------------------------------------------------------------------------- |
| ED visits (3 parts)   | `Urgences_Part1.csv`, `Urgences_Part2.csv`, `Urgences_Part3.csv` | `patient_id`, `visit_id`, `arrival_datetime`, `motif_category`, `triage_level`, `wait_minutes`, `act_unit_cost`, `nb_acts` |
| Patients              | `Patient.xlsx`                                                   | `patientID`, `sex`, `birth_date`, `dept_code`, `city`, `insurance_type`                                                    |
| Departments / Regions | `Department.csv`                                                 | `Region_Name`, `Region_Code`, `Department_Name`, `Department_Code`                                                         |

`Visits.patient_id` → `Patient.patientID` and `Patient.dept_code` → `Department.Department_Code`
